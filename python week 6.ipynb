```python
#function iwth parameter
def add(a,b):
    return a+b
print(add(3,5))
```

```python
#Function with default argument
def power(base,exp=2):
    return base ** exp
print(power(4))
print(power(4,3))
```

```python
#function returning multiple values
def stats(numbers):
    return min(numbers),max(numbers),sum(numbers)/len(numbers)
print(stats([2,4,6,8]))
```

```python
#lambda function
sqaure=lambda x:x**2
print(sqaure(7))
```

```python
def process(data,callback):
    result=data*2
    callback(result)
def show_output(value):
    print("processed:",value)
process(7,show_output)
```
