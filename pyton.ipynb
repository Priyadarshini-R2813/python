```python
# Q9. Factorial using Function
def factorial(n):
    fact = 1
    for i in range(1, n+1):
        fact *= i
    return fact
num = int(input("Enter number "))
print("Factorial:", factorial(num))
```

```python
# Q10. Prime Number Checker
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True
num = int(input("Enter number: "))
if is_prime(num):
    print("Prime Number")
else:
    print("Not Prime")
```

```python
# Q15. Word Frequency Counter
sentence = input("Enter sentence: ")
words = sentence.split()
frequency = {}
for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1
print(frequency)
```

```python
# Q17. Student Result Analyzer
marks = []
for i in range(5):
    mark = int(input("Enter marks: "))
    marks.append(mark)
def average_marks(marks):
    return sum(marks) / len(marks)
def highest_marks(marks):
    return max(marks)
def lowest_marks(marks):
    return min(marks)
def passed_students(marks):
    count = 0
    for mark in marks:
        if mark > 40:
            count += 1
    return count
print("Average:", average_marks(marks))
print("Highest:", highest_marks(marks))
print("Lowest:", lowest_marks(marks))
print("Passed Students:", passed_students(marks))
```

```python
# Q18. Password Strength Checker
def check_password(password):
    has_digit = any(char.isdigit() for char in password)
    has_upper = any(char.isupper() for char in password)
    has_lower = any(char.islower() for char in password)
    min_length = len(password) >= 8
    if has_digit and has_upper and has_lower and min_length:
        return "Strong Password"
    else:
        return "Weak Password"
password = input("Enter password: ")
print(check_password(password))
```

```python
# Q19. Contact Book Application
contacts = {}
def add_contact(name, number):
    contacts[name] = number
def search_contact(name):
    if name in contacts:
        print(name, contacts[name])
    else:
        print("Contact not found")
def delete_contact(name):
    if name in contacts:
        del contacts[name]
def display_contacts():
    print(contacts)
add_contact("Priya", "9876543210")
search_contact("Priya")
display_contacts()
delete_contact("Priya")
display_contacts()
```

```python
# Q20. Mini Banking System
balance = 1000
def deposit(amount):
    global balance
    balance += amount
    print("Balance after deposit:", balance)
def withdraw(amount):
    global balance
    if amount <= balance:
        balance -= amount
        print("Balance after withdrawal:", balance)
    else:
        print("Insufficient Balance")
def check_balance():
    print("Current Balance:", balance)
deposit(500)
withdraw(300)
check_balance()
```

```python
# Q21. HackerRank Style Problem
numbers = [10, 20, 30, 40, 50]
# Find Maximum
maximum = numbers[0]
for num in numbers:
    if num > maximum:
        maximum = num
# Find Minimum
minimum = numbers[0]
for num in numbers:
    if num < minimum:
        minimum = num
# Reverse List
reversed_list = []
for i in range(len(numbers)-1, -1, -1):
    reversed_list.append(numbers[i])
print("Maximum:", maximum)
print("Minimum:", minimum)
print("Reversed List:", reversed_list)
```
